# India AI/ML Job Market — Analysis Notebook\nThis notebook explores the Notion tracker snapshot. It is a personal sample, not a census of the Indian job market. Verify each vacancy before applying.\n\nOpen with Google Colab using the GitHub notebook link in the project README.

```python
import json, urllib.request\nimport pandas as pd\nimport matplotlib.pyplot as plt\nDATA_URL='https://manjunadhgude.github.io/india-tech-jobs/data/jobs.json'\nwith urllib.request.urlopen(DATA_URL) as f: payload=json.load(f)\ndf=pd.DataFrame(payload['records'])\nprint('Snapshot:',payload['generated_at'],'Records:',len(df))\ndisplay(df.head())
```

## Data quality\nInspect missingness and duplicate role-company pairs.

```python
quality=pd.DataFrame({'missing':df.isna().sum(),'missing_pct':(df.isna().mean()*100).round(1)})\ndisplay(quality.sort_values('missing_pct',ascending=False))\nprint('Duplicate role-company pairs:',df.duplicated(['role','company']).sum())
```

## Role category mix

```python
counts=df['category'].fillna('Unclassified').value_counts().head(12).sort_values()\nax=counts.plot(kind='barh',figsize=(10,6),title='Tracked roles by category')\nax.set_xlabel('Number of records'); plt.tight_layout(); plt.show()
```

## Location distribution

```python
counts=df['location'].fillna('Not specified').value_counts().head(15).sort_values()\nax=counts.plot(kind='barh',figsize=(10,7),title='Tracked roles by stated location')\nax.set_xlabel('Number of records'); plt.tight_layout(); plt.show()
```

## Status and compensation\nNot disclosed does not mean unpaid. Status text may contain old verification dates.

```python
status=df['status'].fillna('Not recorded').value_counts().head(12).sort_values()\nax=status.plot(kind='barh',figsize=(10,6),title='Recorded status labels')\nax.set_xlabel('Number of records'); plt.tight_layout(); plt.show()\ndf['compensation_disclosed']=~df['compensation'].fillna('').str.contains('not disclosed',case=False)\ndisplay(df['compensation_disclosed'].value_counts())
```

## Personal fit scores\nScores are available for only some records and are not hiring probabilities.

```python
scores=pd.to_numeric(df['fit_score'],errors='coerce').dropna()\nprint('Scored records:',len(scores),'of',len(df))\ndisplay(scores.describe().round(1))\nif len(scores): scores.plot(kind='hist',bins=10,figsize=(8,4),title='Personal fit-score distribution'); plt.show()
```
